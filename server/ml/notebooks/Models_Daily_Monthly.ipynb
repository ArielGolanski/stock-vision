```python
# ==========================================
# Enhanced LSTM Stock Prediction Model
# Daily & Monthly Change % with Multiple Features
# ==========================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import math
import os
from datetime import datetime, timedelta
import tensorflow as tf
from keras.models import Sequential
from keras.layers import Dense, LSTM, Dropout
from sklearn.preprocessing import MinMaxScaler
import yfinance as yf
import joblib
import warnings
warnings.filterwarnings('ignore')

# ---------------------------
# 1) Load Data with Multiple Features
# ---------------------------
def load_enhanced_data(ticker, start_date="2015-01-01"):
    """Load stock data from a local CSV file and compute enhanced features"""
    # Construct the file path from the ticker symbol
    file_path = f"../data/{ticker}.csv"

    # Check if the data file exists
    if not os.path.exists(file_path):
        print(f"Error: Data file not found at '{file_path}'")
        return pd.DataFrame()  # Return an empty DataFrame if file is not found

    # Load data from the specified CSV file
    df = pd.read_csv(file_path, skiprows=[1, 2])

    df.rename(columns={'Price': 'Date'}, inplace=True)

    # --- Data Cleaning and Formatting ---
    # Standardize column names (e.g., 'date' -> 'Date')
    df.columns = [col.capitalize() for col in df.columns]

    # Convert 'Date' column to datetime objects for filtering and plotting
    df['Date'] = pd.to_datetime(df['Date'])

    # Sort by date to ensure correct order for time-series calculations
    df = df.sort_values('Date').reset_index(drop=True)

    # Filter data to start from the specified start_date
    df = df[df['Date'] >= pd.to_datetime(start_date)].copy()
    
    # --- Feature Engineering (same as before) ---
    # Basic OHLCV features
    df['Daily Change %'] = df['Close'].pct_change() * 100
    df['Volume Change %'] = df['Volume'].pct_change() * 100

    # Technical indicators
    df['SMA_10'] = df['Close'].rolling(window=10).mean()
    df['SMA_30'] = df['Close'].rolling(window=30).mean()
    df['EMA_12'] = df['Close'].ewm(span=12).mean()
    df['EMA_26'] = df['Close'].ewm(span=26).mean()

    # Price-based features
    df['High_Low_Pct'] = ((df['High'] - df['Low']) / df['Close']) * 100
    df['Open_Close_Pct'] = ((df['Close'] - df['Open']) / df['Open']) * 100

    # Volatility
    df['Volatility'] = df['Daily Change %'].rolling(window=20).std()

    # MACD
    df['MACD'] = df['EMA_12'] - df['EMA_26']
    df['MACD_Signal'] = df['MACD'].ewm(span=9).mean()
    df['MACD_Histogram'] = df['MACD'] - df['MACD_Signal']

    # RSI (Relative Strength Index)
    delta = df['Close'].diff()
    gain = (delta.where(delta > 0, 0)).rolling(window=14).mean()
    loss = (-delta.where(delta < 0, 0)).rolling(window=14).mean()
    rs = gain / loss
    df['RSI'] = 100 - (100 / (1 + rs))

    # Monthly change %
    df['Monthly Change %'] = df['Close'].pct_change(periods=21) * 100

    # Drop rows with NaN values created during feature calculation
    df = df.dropna().reset_index(drop=True)

    return df

def prepare_features(df, target_column):
    """Prepare feature matrix for training"""
    feature_columns = [
        'Daily Change %', 'Volume Change %', 'High_Low_Pct', 'Open_Close_Pct',
        'Volatility', 'MACD', 'MACD_Signal', 'MACD_Histogram', 'RSI'
    ]

    # Add monthly change % as feature if not predicting it
    if target_column != 'Monthly Change %':
        feature_columns.append('Monthly Change %')

    features = df[feature_columns].values
    target = df[[target_column]].values

    return features, target, feature_columns

class EnhancedLSTMPredictor:
    def __init__(self, sample_size=60, lstm_units=[128, 64], dropout_rate=0.2):
        self.sample_size = sample_size
        self.lstm_units = lstm_units
        self.dropout_rate = dropout_rate
        self.feature_scaler = MinMaxScaler(feature_range=(0, 1))
        self.target_scaler = MinMaxScaler(feature_range=(0, 1))
        self.model = None

    def build_model(self, n_features):
        """Build enhanced LSTM model"""
        self.model = Sequential([
            LSTM(self.lstm_units[0], return_sequences=True,
                 input_shape=(self.sample_size, n_features)),
            Dropout(self.dropout_rate),
            LSTM(self.lstm_units[1], return_sequences=False),
            Dropout(self.dropout_rate),
            Dense(50, activation='relu'),
            Dropout(self.dropout_rate),
            Dense(25, activation='relu'),
            Dense(1, activation='linear')
        ])

        self.model.compile(optimizer='adam', loss='mean_squared_error', metrics=['mae'])
        return self.model

    def prepare_data(self, features, target):
        """Prepare training and testing datasets"""
        # Scale features and target separately
        scaled_features = self.feature_scaler.fit_transform(features)
        scaled_target = self.target_scaler.fit_transform(target)

        # Split data
        training_len = int(np.ceil(len(scaled_features) * 0.8))

        # Create sequences
        x_train, y_train = self._create_sequences(
            scaled_features[:training_len],
            scaled_target[:training_len]
        )

        x_test, y_test = self._create_sequences(
            scaled_features[training_len-self.sample_size:],
            scaled_target[training_len-self.sample_size:]
        )

        # Adjust y_test to match the correct indices
        y_test_actual = target[training_len:]

        return (x_train, y_train), (x_test, y_test), y_test_actual, training_len

    def _create_sequences(self, features, target):
        """Create sequences for LSTM training"""
        x, y = [], []
        for i in range(self.sample_size, len(features)):
            x.append(features[i-self.sample_size:i])
            y.append(target[i, 0])
        return np.array(x), np.array(y)

    def train(self, x_train, y_train, epochs=20, batch_size=32, validation_split=0.1):
        """Train the model"""
        history = self.model.fit(
            x_train, y_train,
            batch_size=batch_size,
            epochs=epochs,
            validation_split=validation_split,
            verbose=1
        )
        return history

    def predict(self, x_test):
        """Make predictions"""
        pred_scaled = self.model.predict(x_test, verbose=0)

        # Create dummy array for inverse transform
        dummy_array = np.zeros((len(pred_scaled), 1))
        dummy_array[:, 0] = pred_scaled[:, 0]

        predictions = self.target_scaler.inverse_transform(dummy_array)[:, 0]
        return predictions

    def forecast_future(self, df, features, target_column, n_days=5):
        """Forecast future values"""
        # Get the last sequence
        last_features = self.feature_scaler.transform(features[-self.sample_size:])
        last_sequence = last_features.reshape(1, self.sample_size, -1)

        future_predictions = []
        current_price = df['Close'].iloc[-1]
        future_prices = []

        for i in range(n_days):
            # Predict next change %
            next_pred_scaled = self.model.predict(last_sequence, verbose=0)

            # Inverse transform to get actual % change
            dummy = np.zeros((1, 1))
            dummy[0, 0] = next_pred_scaled[0, 0]
            next_change = self.target_scaler.inverse_transform(dummy)[0, 0]

            future_predictions.append(float(next_change))

            # Calculate future price based on change %
            if target_column == 'Daily Change %':
                current_price = current_price * (1 + next_change/100.0)
            elif target_column == 'Monthly Change %':
                # For monthly, we assume the change is spread over ~21 days
                daily_change = next_change / 21.0
                current_price = current_price * (1 + daily_change/100.0)

            future_prices.append(float(current_price))

            # Update sequence for next prediction
            # This is simplified - in reality, you'd need to update all features
            last_sequence = np.roll(last_sequence, -1, axis=1)
            last_sequence[0, -1, 0] = next_pred_scaled[0, 0]  # Update only the first feature

        return future_predictions, future_prices

    def save(self, model_path="model", prefix="daily"):
        """Save model + scalers"""
        # Save keras model
        self.model.save(f"{model_path}_{prefix}.keras")

        # Save scalers
        joblib.dump(self.feature_scaler, f"{model_path}_{prefix}_feature_scaler.pkl")
        joblib.dump(self.target_scaler, f"{model_path}_{prefix}_target_scaler.pkl")
        print(f"Model and scalers saved as {model_path}_{prefix}.*")

    def load(self, model_path="model", prefix="daily"):
        """Load model + scalers"""
        self.model = tf.keras.models.load_model(f"{model_path}_{prefix}.keras")
        self.feature_scaler = joblib.load(f"{model_path}_{prefix}_feature_scaler.pkl")
        self.target_scaler = joblib.load(f"{model_path}_{prefix}_target_scaler.pkl")
        print(f"Model and scalers loaded from {model_path}_{prefix}.*")

def evaluate_model(true_values, predictions, model_name):
    """Evaluate model performance"""
    mse = np.mean((true_values - predictions) ** 2)
    rmse = np.sqrt(mse)
    mae = np.mean(np.abs(true_values - predictions))

    print(f"\n{model_name} Performance Metrics:")
    print(f"MSE: {mse:.4f}")
    print(f"RMSE: {rmse:.4f}")
    print(f"MAE: {mae:.4f}")

    return {'MSE': mse, 'RMSE': rmse, 'MAE': mae}

def plot_results(df, predictions, true_values, future_predictions, future_prices,
                target_column, training_len, ticker, n_display=60):
    """Plot comprehensive results"""
    plt.figure(figsize=(20, 15))

    # Plot 1: Actual vs Predicted Change %
    plt.subplot(3, 2, 1)
    display_len = min(n_display, len(predictions))
    plt.plot(true_values[-display_len:], label=f'Actual {target_column}', alpha=0.7)
    plt.plot(predictions[-display_len:], label=f'Predicted {target_column}', alpha=0.7)
    plt.title(f'{target_column} Prediction - Last {display_len} Days')
    plt.xlabel('Days')
    plt.ylabel('Change %')
    plt.legend()
    plt.grid(True)

    # Plot 2: Price reconstruction and forecast
    plt.subplot(3, 2, 2)

    # Reconstruct prices from predictions
    last_known_price = df['Close'].iloc[training_len - 1]
    pred_prices = []
    price = last_known_price

    for chg in predictions:
        if target_column == 'Daily Change %':
            price = price * (1 + chg/100.0)
        elif target_column == 'Monthly Change %':
            daily_chg = chg / 21.0  # Spread monthly change over ~21 days
            price = price * (1 + daily_chg/100.0)
        pred_prices.append(price)

    true_prices = df['Close'].values[training_len:]

    display_len = min(50, len(pred_prices))
    plt.plot(range(display_len), true_prices[-display_len:],
             label='Actual Price', linewidth=2)
    plt.plot(range(display_len), pred_prices[-display_len:],
             label='Reconstructed Price', linewidth=2, alpha=0.8)
    plt.plot(range(display_len, display_len + len(future_prices)), future_prices,
             marker='o', linestyle='--', label='Future Forecast', linewidth=2)
    plt.title(f'{ticker} Price Prediction ({target_column} Model)')
    plt.xlabel('Days')
    plt.ylabel('Price ($)')
    plt.legend()
    plt.grid(True)

    # Plot 3: Training Loss (if history available)
    plt.subplot(3, 2, 3)
    plt.text(0.1, 0.5, f'Model Features:\n• Enhanced Technical Indicators\n• Volume Analysis\n• MACD & RSI\n• Volatility Measures',
             transform=plt.gca().transAxes, fontsize=12, verticalalignment='center',
             bbox=dict(boxstyle="round,pad=0.3", facecolor="lightblue", alpha=0.5))
    plt.title('Model Information')
    plt.axis('off')

    # Plot 4: Feature correlation heatmap (simplified)
    plt.subplot(3, 2, 4)
    recent_data = df[['Daily Change %', 'Volume Change %', 'RSI', 'MACD']].tail(100)
    correlation_matrix = recent_data.corr()
    im = plt.imshow(correlation_matrix, cmap='coolwarm', aspect='auto')
    plt.colorbar(im)
    plt.xticks(range(len(correlation_matrix.columns)), correlation_matrix.columns, rotation=45)
    plt.yticks(range(len(correlation_matrix.columns)), correlation_matrix.columns)
    plt.title('Feature Correlation Matrix')

    # Plot 5: Residuals
    plt.subplot(3, 2, 5)
    residuals = true_values - predictions
    plt.scatter(range(len(residuals)), residuals, alpha=0.6)
    plt.axhline(y=0, color='r', linestyle='--')
    plt.title('Prediction Residuals')
    plt.xlabel('Sample')
    plt.ylabel('Residual')
    plt.grid(True)

    # Plot 6: Future predictions details
    plt.subplot(3, 2, 6)
    days = list(range(1, len(future_predictions) + 1))
    bars = plt.bar(days, future_predictions, alpha=0.7)
    for i, (day, pred) in enumerate(zip(days, future_predictions)):
        plt.text(day, pred + (0.01 if pred >= 0 else -0.01), f'{pred:.2f}%',
                ha='center', va='bottom' if pred >= 0 else 'top')
    plt.title(f'Future {target_column} Predictions')
    plt.xlabel('Days Ahead')
    plt.ylabel('Predicted Change %')
    plt.grid(True)

    plt.tight_layout()
    plt.show()
# ---------------------------
# Main Execution
# ---------------------------
def main(tickers=None, n_days=5):
    if tickers is None:
        tickers = ["MSFT", "IBM", "T"]  # default tickers

    for ticker in tickers:
        print(f"\n\n{'#'*80}")
        print(f"Processing ticker: {ticker}")
        print(f"{'#'*80}\n")

        print(f"Loading enhanced data for {ticker}...")
        df = load_enhanced_data(ticker)
        print(f"Loaded {len(df)} days of data with enhanced features")

        # Train models for both daily and monthly predictions
        for target_column in ['Daily Change %', 'Monthly Change %']:
            print(f"\n{'='*60}")
            print(f"Training model for {ticker} → {target_column}")
            print(f"{'='*60}")

            # Prepare data
            features, target, feature_names = prepare_features(df, target_column)
            print(f"Using {len(feature_names)} features: {feature_names}")

            # Initialize and train model
            predictor = EnhancedLSTMPredictor(sample_size=60, lstm_units=[128, 64])
            predictor.build_model(n_features=features.shape[1])

            (x_train, y_train), (x_test, y_test), y_test_actual, training_len = predictor.prepare_data(features, target)

            print(f"Training data shape: {x_train.shape}")
            print(f"Training model...")

            history = predictor.train(x_train, y_train, epochs=15, batch_size=32)

            # Ensure ml_models/<ticker>/ directory exists
            model_dir = f"../ml_models/{ticker}/{ticker}_enhanced_lstm"
            os.makedirs(model_dir, exist_ok=True)

            # Save trained model + scalers
            prefix = "daily" if target_column == "Daily Change %" else "monthly"
            predictor.save(model_path=model_dir, prefix=prefix)

            # Make predictions
            predictions = predictor.predict(x_test)

            # Evaluate model
            metrics = evaluate_model(y_test_actual.flatten(), predictions, f"{ticker} {target_column} Model")

            # Future forecasting
            future_changes, future_prices = predictor.forecast_future(df, features, target_column, n_days=n_days)

            print(f"\nFuture {target_column} Predictions for {ticker}:")
            for i, (change, price) in enumerate(zip(future_changes, future_prices), 1):
                print(f"Day {i}: {change:.2f}% change → Price: ${price:.2f}")

            # Plot results
            plot_results(
                df, predictions, y_test_actual.flatten(),
                future_changes, future_prices,
                target_column, training_len, ticker
            )

if __name__ == "__main__":
    main(tickers=["MSFT", "IBM", "T", "AAPL"], n_days=5)
```
